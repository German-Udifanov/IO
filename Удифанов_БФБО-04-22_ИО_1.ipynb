## Занятие 1. Производственные функции. Экстремум функции нескольких переменных.

##  * Градиентный спуск, метод градиентного спуска — численный метод нахождения локального минимума или максимума функции с помощью движения вдоль градиента

1. Дана производственная функция вида: 

𝑥, 𝑦 – переменные, определяющие количество использованных ресурсов.
Найти экстремум функции и параметры, при которых он достигается.


```python
import numpy as np

# определяем функцию и её градиент
def Q(point):
    x, y = point
    return 2*x**2 + 3*x*y + 4*y**2 - 6*x - 12*y + 10

# вычисление градиента 
def gradient(point):
    x, y = point
    pd_x = 4*x + 3*y - 6  # частная производная по x
    pd_y = 3*x + 8*y - 12  # частная производная по y
    return np.array([pd_x, pd_y]) # возвращает градиент в виде массива NumPy, содержащего две частные производные

# метод градиентного спуска
def gradient_descent(starting_point, learning_rate, num_iterations):
    point = np.array(starting_point) # преобразует starting_point в массив для удобства операций
    for _ in range(num_iterations):  # цикл для обновления точки
        grad = gradient(point)
        point = point - learning_rate * grad  # обновление точки
    return point

# параметры метода
starting_point = [0, 0]  # начальная точка
learning_rate = 0.01  # скорость обучения
num_iterations = 10000  # количество итераций

# запуск метода градиентного спуска
minimum_point = gradient_descent(starting_point, learning_rate, num_iterations) # значение экстремума
minimum_value = Q(minimum_point) # значение функции в точке экстремума

# вывод результатов
print(f"Минимум достигается в точке: {minimum_point}") 
print(f"Значение функции в этой точке: {minimum_value}")

# вычисляем вторые производные
def second_derivatives(point):
    x, y = point
    pd_x2 = 4  # вторая производная по x
    pd_y2 = 8  # вторая производная по y
    return pd_x2, pd_y2

# проверяем тип экстремума
pd_x2, pd_y2 = second_derivatives(minimum_point)

if pd_x2 > 0 and pd_y2 > 0:
    print("Найдена точка минимума.")
else:
    print("Найдена точка максимума.")
```

2. Дана производственная функция вида:

𝑥, 𝑦 – переменные, определяющие количество использованных ресурсов.
Найти экстремум функции и параметры, при которых он достигается.

```python
import numpy as np

# определяем функцию и её градиент
def Q(point):
    x, y = point
    return x**2 + y**2 - 4*x - 2*y + 5

def gradient(point):
    x, y = point
    pd_x = 2*x - 4  # частная производная по x
    pd_y = 2*y - 2  # частная производная по y
    return np.array([pd_x, pd_y])

# метод градиентного спуска
def gradient_descent(starting_point, learning_rate, num_iterations):
    point = np.array(starting_point)
    for _ in range(num_iterations):
        grad = gradient(point)
        point = point - learning_rate * grad  # обновление точки
    return point

# параметры метода
starting_point = [0, 0]  # начальная точка
learning_rate = 0.01  # скорость обучения
num_iterations = 10000  # количество итераций

# запуск метода градиентного спуска
minimum_point = gradient_descent(starting_point, learning_rate, num_iterations)
minimum_value = Q(minimum_point)

# вывод результатов
print(f"Минимум достигается в точке: {minimum_point}")
print(f"Значение функции в этой точке: {minimum_value}")

# проверяем тип экстремума
pd_x2 = 2  # вторая производная по x
pd_y2 = 2  # вторая производная по y

if pd_x2 > 0 and pd_y2 > 0:
    print("Найдена точка минимума.")
else:
    print("Найдена точка максимума или седловая точка.")
```
