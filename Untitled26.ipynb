```python
import math

def calculate_eoq():
    print("Задача 1:")
    # Параметры для задачи 1
    D1 = 120000  # Годовая потребность (детали/год)
    S1 = 10000   # Стоимость заказа (ден. ед./заказ)
    H1 = 0.5 * 365  # Годовая стоимость хранения (ден. ед./деталь/год)
    L1 = 20      # Время выполнения заказа (дни)
    C1 = 5       # Убытки от дефицита (ден. ед./деталь)
    
    # Расчет оптимального размера заказа (EOQ)
    Q1_opt = math.sqrt((2 * D1 * S1) / H1)
    # Расчет оптимального интервала между заказами (в днях)
    T1_opt = (Q1_opt / D1) * 365
    
    # Точка перезаказа (Reorder Point)
    R1 = (D1 / 365) * L1
    
    print(f"Оптимальный размер партии: {Q1_opt:.2f} деталей")
    print(f"Оптимальный интервал между поставками: {T1_opt:.2f} дней")
    print(f"Точка перезаказа: {R1:.2f} деталей (когда остаток достигает этого уровня)")
    
    print("\nЗадача 2:")
    # Параметры для задачи 2
    D2 = 15000 * 12  # Годовая потребность (детали/год)
    S2 = 1000    # Стоимость заказа (ден. ед./заказ)
    H2 = 1 * 365 # Годовая стоимость хранения (ден. ед./деталь/год)
    L2 = 3       # Время выполнения заказа (дни)
    C2 = 5       # Убытки от дефицита (ден. ед./деталь)
    
    # Расчет оптимального размера заказа (EOQ)
    Q2_opt = math.sqrt((2 * D2 * S2) / H2)
    # Расчет оптимального интервала между заказами (в днях)
    T2_opt = (Q2_opt / D2) * 365
    
    # Точка перезаказа (Reorder Point)
    R2 = (D2 / 365) * L2
    
    print(f"Оптимальный размер партии: {Q2_opt:.2f} деталей")
    print(f"Оптимальный интервал между поставками: {T2_opt:.2f} дней")
    print(f"Точка перезаказа: {R2:.2f} деталей (когда остаток достигает этого уровня)")

# Выполняем расчеты
calculate_eoq()
```
