```python
from sympy import symbols, diff, Eq, solve, Matrix

def lagrange_method(f, constraint_eq, variables):
    """
    Функция для нахождения экстремума методом множителей Лагранжа
    
    Параметры:
    f - целевая функция
    constraint_eq - уравнение ограничения (равно 0)
    variables - список переменных
    
    Возвращает:
    Решение системы уравнений
    """
    # Создаем символ для множителя Лагранжа
    λ = symbols('λ')
    
    # Формируем лагранжиан
    L = f - λ * constraint_eq
    
    # Список всех переменных (включая λ)
    all_vars = list(variables) + [λ]
    
    # Вычисляем частные производные
    equations = []
    for var in all_vars:
        equations.append(Eq(diff(L, var), 0))
    
    # Решаем систему уравнений
    solution = solve(equations, all_vars, dict=True)
    
    return solution

# Задача 1
print("Задача 1:")
x, y = symbols('x y')
f1 = x**2 + y**2 - 4*x - 2*y + 5
constraint1 = x + y - 3  # x + y = 3 → x + y - 3 = 0

solution1 = lagrange_method(f1, constraint1, (x, y))

# Вычисляем значение функции в точке экстремума
if solution1:
    for sol in solution1:
        x_val = sol[x]
        y_val = sol[y]
        f_value = f1.subs({x: x_val, y: y_val})
        print(f"Решение: x = {x_val}, y = {y_val}, λ = {sol[λ]}")
        print(f"Значение функции: f({x_val}, {y_val}) = {f_value}")
        
        # Проверка на минимум/максимум через Гессиан
        H = Matrix([[diff(f1, x, 2), diff(diff(f1, x), y)],
                    [diff(diff(f1, y), x), diff(f1, y, 2)]])
        det_H = H.det()
        d2f_dx2 = H[0, 0]
        
        if det_H > 0 and d2f_dx2 > 0:
            print("Точка является минимумом")
        elif det_H > 0 and d2f_dx2 < 0:
            print("Точка является максимумом")
        else:
            print("Точка является седловой или неопределена")
else:
    print("Решений не найдено")

# Задача 2
print("\nЗадача 2:")
f2 = 3*x**2 + 4*y**2 - 6*x - 8*y + 10
constraint2 = x + y - 5  # x + y = 5 → x + y - 5 = 0

solution2 = lagrange_method(f2, constraint2, (x, y))

if solution2:
    for sol in solution2:
        x_val = sol[x]
        y_val = sol[y]
        f_value = f2.subs({x: x_val, y: y_val})
        print(f"Решение: x = {x_val}, y = {y_val}, λ = {sol[λ]}")
        print(f"Значение функции: f({x_val}, {y_val}) = {f_value}")
        
        # Проверка на минимум/максимум через Гессиан
        H = Matrix([[diff(f2, x, 2), diff(diff(f2, x), y)],
                    [diff(diff(f2, y), x), diff(f2, y, 2)]])
        det_H = H.det()
        d2f_dx2 = H[0, 0]
        
        if det_H > 0 and d2f_dx2 > 0:
            print("Точка является минимумом")
        elif det_H > 0 and d2f_dx2 < 0:
            print("Точка является максимумом")
        else:
            print("Точка является седловой или неопределена")
else:
    print("Решений не найдено")
```
