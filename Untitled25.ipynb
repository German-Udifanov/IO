```python
import numpy as np
from scipy.integrate import quad

def calculate_warehouse():
    # Задача 1
    print("Задача 1:")
    # Определяем функцию интенсивности для задачи 1
    def intensity1(t):
        if 0 <= t <= 4:  # с 8:00 до 12:00 (4 часа)
            return 100 + (300 - 100)/4 * t  # линейный рост от 100 до 300
        elif 4 < t <= 12:  # с 12:00 до 20:00 (8 часов)
            return 300 - (300 - 50)/8 * (t - 4)  # линейное снижение от 300 до 50
        else:
            return 0
    
    # Интегрируем для первых 30 минут (0.5 часа)
    q1_30min, _ = quad(intensity1, 0, 0.5)
    # Интегрируем за весь рабочий день (12 часов)
    q1_total, _ = quad(intensity1, 0, 12)
    
    print(f"Количество товара через 30 минут: {q1_30min:.2f} единиц")
    print(f"Количество товара в конце смены: {q1_total:.2f} единиц")
    
    # Задача 2
    print("\nЗадача 2:")
    # Определяем функцию интенсивности для задачи 2
    def intensity2(t):
        if 0 <= t <= 4:  # с 8:00 до 12:00 (4 часа)
            return 50 + (200 - 50)/4 * t  # линейный рост от 50 до 200
        elif 4 < t <= 12:  # с 12:00 до 20:00 (8 часов)
            return 200 - (200 - 100)/8 * (t - 4)  # линейное снижение от 200 до 100
        else:
            return 0
    
    # Интегрируем для первых 30 минут (0.5 часа)
    q2_30min, _ = quad(intensity2, 0, 0.5)
    # Интегрируем за весь рабочий день (12 часов)
    q2_total, _ = quad(intensity2, 0, 12)
    
    print(f"Количество товара через 30 минут: {q2_30min:.2f} единиц")
    print(f"Количество товара в конце смены: {q2_total:.2f} единиц")

# Выполняем расчеты
calculate_warehouse()
```
