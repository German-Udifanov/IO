```python
from sympy import symbols, diff, solve, Eq, hessian, Matrix

def find_extremum(Q, x, y):
    # Находим частные производные первого порядка
    dQ_dx = diff(Q, x)
    dQ_dy = diff(Q, y)
    
    # Решаем систему уравнений для нахождения стационарной точки
    critical_point = solve((Eq(dQ_dx, 0), Eq(dQ_dy, 0)), (x, y))
    
    # Создаем Гессиан (матрицу вторых производных)
    H = hessian(Q, (x, y))
    
    # Вычисляем определитель Гессиана и значение Q в критической точке
    if critical_point:
        x_val = critical_point[x]
        y_val = critical_point[y]
        det_H = H.det().subs({x: x_val, y: y_val})
        d2Q_dx2 = H[0, 0].subs({x: x_val, y: y_val})
        
        # Определяем тип экстремума
        if det_H > 0 and d2Q_dx2 > 0:
            extremum_type = "минимум"
        elif det_H > 0 and d2Q_dx2 < 0:
            extremum_type = "максимум"
        else:
            extremum_type = "седловая точка"
        
        Q_value = Q.subs({x: x_val, y: y_val})
        
        return {
            "critical_point": (x_val, y_val),
            "extremum_type": extremum_type,
            "Q_value": Q_value,
            "hessian": H
        }
    else:
        return None

# Задача 1
x, y = symbols('x y')
Q1 = 2*x**2 + 3*x*y + 4*y**2 - 6*x - 12*y + 10
result1 = find_extremum(Q1, x, y)

print("Задача 1:")
print(f"Функция: Q(x, y) = {Q1}")
print(f"Критическая точка: {result1['critical_point']}")
print(f"Тип экстремума: {result1['extremum_type']}")
print(f"Значение функции: {result1['Q_value'].evalf()}")
print(f"Матрица Гессиана:\n{result1['hessian']}\n")

# Задача 2
Q2 = x**2 + y**2 - 4*x - 2*y + 5
result2 = find_extremum(Q2, x, y)

print("Задача 2:")
print(f"Функция: Q(x, y) = {Q2}")
print(f"Критическая точка: {result2['critical_point']}")
print(f"Тип экстремума: {result2['extremum_type']}")
print(f"Значение функции: {result2['Q_value'].evalf()}")
print(f"Матрица Гессиана:\n{result2['hessian']}")
```
